```python
from io import StringIO

import pandas as pd
import requests

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

URL = "https://en.wikipedia.org/wiki/List_of_current_NHL_captains_and_alternate_captains"

response = requests.get(URL, headers=HEADERS)
response.raise_for_status()

tables = pd.read_html(StringIO(response.text))

print(f"{len(tables)} tables found")
```

```python
for i, table in enumerate(tables):

    print("=" * 80)
    print(f"TABLE {i}")

    if isinstance(table.columns, pd.MultiIndex):
        table.columns = [
            c[0] if c[0] == c[1] else "_".join(str(x) for x in c if x)
            for c in table.columns
        ]

    print(table.columns.tolist())
    print(f"{len(table)} rows")
```

```python
from bs4 import BeautifulSoup
import requests

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

URL = "https://en.wikipedia.org/wiki/List_of_current_NHL_captains_and_alternate_captains"

html = requests.get(URL, headers=HEADERS).text

soup = BeautifulSoup(html, "html.parser")

print("Downloaded")
```

```python
tables = soup.find_all("table", class_="wikitable")

print(len(tables))

for i, table in enumerate(tables):
    print("=" * 80)
    print(i)
    print(table.find("caption").get_text(strip=True))
```

```python
alternate_table = tables[2]

rows = alternate_table.find_all("tr")

print(f"{len(rows)} rows")

for i, row in enumerate(rows[:20]):

    cells = row.find_all(["th", "td"])

    print("=" * 80)
    print(f"Row {i} ({len(cells)} cells)")
    print([c.get_text(" ", strip=True) for c in cells])
```

```python
import pandas as pd

records = []

current_team = None
current_alternates = []

rows = alternate_table.find_all("tr")

for row in rows[1:]:      # Skip header

    cells = row.find_all(["td", "th"])

    texts = [
        clean_text(c.get_text(" ", strip=True))
        for c in cells
    ]

    # Ignore blank rows
    if len(texts) == 0:
        continue

    # First team in the table
    if len(texts) == 4:

        if current_team is not None:

            records.append({
                "Team": current_team,
                "Alternates": current_alternates.copy()
            })

        current_team = texts[0]
        current_alternates = [texts[1]]

    # New team heading
    elif len(texts) == 1:

        if current_team is not None:

            records.append({
                "Team": current_team,
                "Alternates": current_alternates.copy()
            })

        current_team = texts[0]
        current_alternates = []

    # Additional alternate captain
    elif len(texts) == 3:

        current_alternates.append(texts[0])

# Last team
records.append({
    "Team": current_team,
    "Alternates": current_alternates.copy()
})

alternate_parsed = pd.DataFrame(records)

alternate_parsed
```

```python
import re

def clean_text(value):

    if value is None:
        return None

    value = str(value)

    # Remove Wikipedia reference numbers
    value = re.sub(r"\[\s*\d+\s*\]", "", value)

    # Remove dagger symbols
    value = value.replace("†", "")

    # Normalise whitespace
    value = " ".join(value.split())

    return value.strip()
```
